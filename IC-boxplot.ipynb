```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
accuracies = pd.read_csv('Datasets/Boxplot-data/models_accuracy_30_each.csv')

sns.set_style('whitegrid', {'axes.grid' : False})
boxplot = sns.boxplot(x='Model', y='Value', data=accuracies, palette='light:#5A9')
plt.ylim(75,90)


plt.xlabel('')
plt.ylabel('Accuracy (%)')
plt.title('')

plt.savefig('boxplot.png', dpi=300, bbox_inches='tight', facecolor='white')
```
