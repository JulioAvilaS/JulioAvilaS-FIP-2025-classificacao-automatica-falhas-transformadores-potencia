```python
import pandas as pd
from enum import Enum
import random
from sklearn.preprocessing import RobustScaler
from sklearn.model_selection import train_test_split
```

## Recebe os arquivos de texto e transforma em listas

```python
filename = "Datasets/Ibrahim-dataset/SMOTE-Oversampled-119-dataset/train_samples.csv"
df = pd.read_csv(filename)
train_samples = df.values.tolist()

filename = "Datasets/Ibrahim-dataset/SMOTE-Oversampled-119-dataset/train_labels.csv"
df = pd.read_csv(filename)
train_labels = df.values.tolist()

filename = "Datasets/Ibrahim-dataset/Gas-80-20-dataset/test_samples.csv"
df = pd.read_csv(filename)
test_samples = df.values.tolist()

filename = "Datasets/Ibrahim-dataset/Gas-80-20-dataset/test_labels.csv"
df = pd.read_csv(filename)
test_labels = df.values.tolist()
```

## Calcula as porcentagens de cada gás

```python
def calculate_gas_percentages(samples):
    """
    Calcula a porcentagem de cada gás em relação ao total de gases.
    """
    gas_percentages = []
    for sample in samples:
        total_gas = sum(sample)
        if total_gas > 0:
            percentages = [gas / total_gas * 100 for gas in sample]
        else:
            percentages = [0] * len(sample)
        gas_percentages.append(percentages)
    return gas_percentages
```

## Calcular as razões dos conjuntos de treino e teste

```python
def calculate_ratios(samples):
    reasons_list = []
    for sample in samples:
        h2, ch4, c2h2, c2h4, c2h6 = sample
        ratios = [
            ch4 / h2 if h2 != 0 else 0,
            c2h6 / ch4 if ch4 != 0 else 0,
            c2h4 / c2h6 if c2h6 != 0 else 0,
            c2h2 / c2h4 if c2h4 != 0 else 0
        ]
        reasons_list.append(ratios)
    return reasons_list

train_reasons = calculate_ratios(train_samples)
test_reasons = calculate_ratios(test_samples)

train_samples = calculate_gas_percentages(train_samples)
test_samples = calculate_gas_percentages(test_samples)
```

## Concatenar as listas de ppm e as listas de razões 

```python
def concatenate_lists(list1, list2):
    df1 = pd.DataFrame(list1)
    df2 = pd.DataFrame(list2)
    resultsdf = pd.concat([df1, df2], axis=1, ignore_index=True)
    return resultsdf.values.tolist()

train_samples_and_reasons = concatenate_lists(train_samples, train_reasons)
test_samples_and_reasons = concatenate_lists(test_samples, test_reasons)
```

## Escrever os arquivos csv dos gases separados:

```python
df = pd.DataFrame(train_samples_and_reasons)
df.columns = ['h2', 'ch4', 'c2h2', 'c2h4', 'c2h6', 'ch4/h2', 'c2h6/ch4', 'c2h4/c2h6', 'c2h2/c2h4']
df.to_csv('Datasets/Ibrahim-dataset/Gas-and-Reasons-dataset/train_samples.csv', index=False)

df = pd.DataFrame(train_labels)
df.columns = ['act']
df.to_csv('Datasets/Ibrahim-dataset/Gas-and-Reasons-dataset/train_labels.csv', index=False)

df = pd.DataFrame(test_samples_and_reasons)
df.columns = ['h2', 'ch4', 'c2h2', 'c2h4', 'c2h6', 'ch4/h2', 'c2h6/ch4', 'c2h4/c2h6', 'c2h2/c2h4']
df.to_csv('Datasets/Ibrahim-dataset/Gas-and-Reasons-dataset/test_samples.csv', index=False)

df = pd.DataFrame(test_labels)
df.columns = ['act']
df.to_csv('Datasets/Ibrahim-dataset/Gas-and-Reasons-dataset/test_labels.csv', index=False)
```
