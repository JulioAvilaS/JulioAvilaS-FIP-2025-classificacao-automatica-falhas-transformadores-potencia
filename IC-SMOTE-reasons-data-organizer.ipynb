```python
import pandas as pd
from enum import Enum
import random
from sklearn.preprocessing import MinMaxScaler
```

## Utilizar a biblioteca pandas para verificar as linhas iguais

```python
filename = 'Datasets/Ibrahim-dataset/Entire-dataset/Ibrahim_original_dataset.csv'
df = pd.read_csv(filename)
original_list = df.values.tolist()

filename = 'Datasets/Ibrahim-dataset/Entire-dataset/not_fail.csv'
df = pd.read_csv(filename)
not_fail_list = df.values.tolist()

samples_list = original_list  + not_fail_list


print("Amostras unificadas:", samples_list)
```

```python
count = 0
duplicates = []
seen = set()

for item in samples_list:
    tuple_item = tuple(item)
    if tuple_item in seen:
        count += 1
        duplicates.append(item)
    else:
        seen.add(tuple_item)

print("\nItens duplicados:", duplicates)
unique_samples = [list(item) for item in seen]
random.shuffle(unique_samples)
```

## Normalizar as falhas antes de dividir:

```python
scaler = MinMaxScaler(feature_range=(0,1))
scaled_train_samples = scaler.fit_transform(unique_samples)
```

## Separa as porcentagens de cada lista:

```python
db_20_percent = int(0.2 * len(unique_samples))

train_list = unique_samples[db_20_percent:]
test_list = unique_samples[:db_20_percent]
```

## Transforma, divide as labels e armazena com os valores corretos:

```python
train_labels_list = []
train_gas_cleaned_samples = []
test_labels_list = []
test_gas_cleaned_samples = []

for row in train_list:
    train_labels_list.append(int(row[-1]) + 1)  # Armazena o último elemento
    gas_cleaned_row = row[:-1]  # Remove o último elemento
    train_gas_cleaned_samples.append(gas_cleaned_row)

for row in test_list:
    test_labels_list.append(int(row[-1]) + 1)  # Armazena o último elemento
    gas_cleaned_row = row[:-1]  # Remove o último elemento
    test_gas_cleaned_samples.append(gas_cleaned_row)


print("\nNúmero de amostras únicas:", len(train_gas_cleaned_samples) + len(test_gas_cleaned_samples))
```

## Reorganizar a ordem dos gases:
### A ordem foi alterada para se encaixar no padrão pré estabelecido pelo projeto <br/> (h2, ch2, c2h2, c2h4, c2h6).

```python
new_order = [0,1,4,3,2]

train_gas_reorganized_samples = [tuple(tupla[i] for i in new_order) for tupla in train_gas_cleaned_samples]
test_gas_reorganized_samples = [tuple(tupla[i] for i in new_order) for tupla in test_gas_cleaned_samples]

print(train_gas_reorganized_samples)
print(test_gas_reorganized_samples)
```

## Realiza as divisões:

```python
class GasType(Enum):
    H2 = 0
    CH4 = 1
    C2H2 = 2
    C2H4 = 3
    C2H6 = 4

def calculate_reasons(sample):
    decimal_places = 4
    return[
        round(sample[GasType.C2H2.value] / sample[GasType.C2H4.value], decimal_places), # c2h2/c2h4
        round(sample[GasType.CH4.value] / sample[GasType.H2.value], decimal_places), # ch4/h2
        round(sample[GasType.C2H4.value] / sample[GasType.C2H6.value], decimal_places) # c2h4/c2h6
    ]

train_samples_list = [calculate_reasons(sample) for sample in train_gas_reorganized_samples]
test_samples_list = [calculate_reasons(sample) for sample in test_gas_reorganized_samples]
```

## Escrever os arquivos csv dos gases separados:

```python
df = pd.DataFrame(train_gas_reorganized_samples)
df.columns = ['h2', 'ch4', 'c2h2', 'c2h4', 'c2h6']
df.to_csv('Datasets/Ibrahim-dataset/Normalized-80-20-dataset/train_samples.csv', index=False)

df = pd.DataFrame(train_labels_list)
df.columns = ['act']
df.to_csv('Datasets/Ibrahim-dataset/Normalized-80-20-dataset/train_labels.csv', index=False)

df = pd.DataFrame(test_gas_reorganized_samples)
df.columns = ['h2', 'ch4', 'c2h2', 'c2h4', 'c2h6']
df.to_csv('Datasets/Ibrahim-dataset/Normalized-80-20-dataset/test_samples.csv', index=False)

df = pd.DataFrame(test_labels_list)
df.columns = ['act']
df.to_csv('Datasets/Ibrahim-dataset/Normalized-80-20-dataset/test_labels.csv', index=False)
```

## Escrever os arquivos csv das razões:

```python
df = pd.DataFrame(train_samples_list)
df.columns = ['c2h2/c2h4','ch4/h2','c2h4/c2h6']
df.to_csv('Datasets/Ibrahim-dataset/Reasons-80-20-dataset/train_samples.csv', index=False)

df = pd.DataFrame(train_labels_list)
df.columns = ['act']
df.to_csv('Datasets/Ibrahim-dataset/Reasons-80-20-dataset/train_labels.csv', index=False)

df = pd.DataFrame(test_samples_list)
df.columns = ['c2h2/c2h4','ch4/h2','c2h4/c2h6']
df.to_csv('Datasets/Ibrahim-dataset/Reasons-80-20-dataset/test_samples.csv', index=False)

df = pd.DataFrame(test_labels_list)
df.columns = ['act']
df.to_csv('Datasets/Ibrahim-dataset/Reasons-80-20-dataset/test_labels.csv', index=False)
```
