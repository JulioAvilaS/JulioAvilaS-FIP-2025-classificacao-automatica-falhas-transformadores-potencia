```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
```

```python

samplesCount1 = [59, 35, 59, 95, 72, 35, 59]
samplesCount2 = [95, 95, 95, 95, 95, 95, 95]
labels = ['NF', 'PD', 'D1', 'D2', 'T1', 'T2', 'T3']

x = np.arange(len(labels))  
width = 0.35  

plt.bar(x - width/2, samplesCount1, width, color='tomato', label='Unbalanced database')
plt.bar(x + width/2, samplesCount2, width, color='royalblue', label='Balanced database')

plt.xticks(x, labels)
plt.ylabel('Number of samples')
plt.title('Comparison between databases')

plt.legend()
plt.grid(axis='y', linestyle='--', alpha=0.5)
plt.tight_layout()

plt.savefig('histogramaBalanceamento.png', dpi=300, bbox_inches='tight', facecolor='white')
plt.show()
```
