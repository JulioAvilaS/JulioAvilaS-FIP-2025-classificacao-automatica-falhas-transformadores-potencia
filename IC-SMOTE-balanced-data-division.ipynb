```python
import pandas as pd
import random 
import numpy as np
from sklearn.preprocessing import MinMaxScaler, RobustScaler
from sklearn.model_selection import train_test_split
```

```python
filename = 'Datasets/Ibrahim-dataset/Oversampled-119-dataset/entire_dataset.csv'
df = pd.read_csv(filename)
original_oversampled_dataset = df.values.tolist()
random.shuffle(original_oversampled_dataset)
```

```python
unbalanced_df = pd.DataFrame(original_oversampled_dataset)
print(unbalanced_df.columns.tolist())

unbalanced_samples = unbalanced_df.drop(columns=[5])
labels = unbalanced_df[5]
```

```python
# Supondo que train_samples já esteja carregado
scaler = RobustScaler()
balanced_samples = scaler.fit_transform(unbalanced_samples)

balanced__rounded_samples = np.round(balanced_samples, decimals=6)
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    balanced__rounded_samples,
    labels,
    test_size=0.2,
    random_state=42,
    stratify=labels
)
```

```python
df = pd.DataFrame(X_train)
df.columns = ['h2', 'ch4', 'c2h2', 'c2h4', 'c2h6']
df.to_csv('Datasets/Ibrahim-dataset/Oversampled-119-dataset/train_samples.csv', index=False)

df = pd.DataFrame(y_train)
df.columns = ['act']
df.to_csv('Datasets/Ibrahim-dataset/Oversampled-119-dataset/train_labels.csv', index=False)

df = pd.DataFrame(X_test)
df.columns = ['h2', 'ch4', 'c2h2', 'c2h4', 'c2h6']
df.to_csv('Datasets/Ibrahim-dataset/Oversampled-119-dataset/test_samples.csv', index=False)

df = pd.DataFrame(y_test)
df.columns = ['act']
df.to_csv('Datasets/Ibrahim-dataset/Oversampled-119-dataset/test_labels.csv', index=False)
```
